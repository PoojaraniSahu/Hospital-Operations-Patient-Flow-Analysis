```python
import seaborn as sns
import numpy as np
import pandas as pd  
import matplotlib.pyplot as plt
```

```python
df=pd.read_excel('Hospital_data.xlsx')
print(df)
```

```python
#Q1Trim and standardize patient, encounter, appointment, and order IDs; map aliases using approved master tables.
# Columns containing IDs
id_columns = ["Patient ID", "Encounter ID", "Appointment ID", "Lab Order ID"]

# Trim spaces and convert IDs to uppercase
for col in id_columns:
    df[col] = df[col].astype("string").str.strip().str.upper()

# Display the cleaned IDs
print(df[id_columns].head(10))
```

```python
#2. Replace direct patient identifiers in analytics extracts with a consistent pseudonymous key.
# Create a pseudonymous patient key
df["Patient_Key"] = df["Patient ID"].astype("string").str.strip().str.upper()
# Display the output
print(df[["Patient_Key", "Encounter ID", "Facility", "Dept"]].head(10))
```

```python
#3. Parse timestamps and normalize them to the facility's documented timezone or UTC.
time_columns = ["Arrival", "Triage", "Discharge", "Appointment DateTime",
                "Admission", "Transfer", "Procedure DateTime",
                "Lab Result DateTime", "Shift Start", "Shift End"]

for col in time_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce", utc=True)
df[time_columns].head(10)
```

```python
#4. Standardize facility, ward, department, specialty, and clinician role names using reference lists.
mappings = {
    "Facility": {
        "NORTH HOSP": "NORTH-HOSPITAL",
        "NORTH HOSPITAL": "NORTH-HOSPITAL",
        "SOUTH HOSPITAL": "SOUTH-HOSPITAL"
    },
    "Ward": {
        "ER WARD": "EMERGENCY WARD",
        "ICU-1": "ICU"
    },
    "Dept": {
        "ER": "EMERGENCY",
        "CARDIO": "CARDIOLOGY",
        "LAB": "LABORATORY"
    },
    "Specialty": {
        "CARDIO": "CARDIOLOGY",
        "PEDS": "PEDIATRICS"
    },
    "Clinician Role": {
        "MD": "PHYSICIAN",
        "DOCTOR": "PHYSICIAN"
    }
}

columns = ["Facility", "Ward", "Dept", "Specialty", "Clinician Role"]

for col in columns:
    df[col] = df[col].astype("string").str.strip().str.upper()
    df[col] = df[col].replace(mappings[col])

df[columns].head(10)
```

```python
#5. Normalize appointment and encounter statuses (scheduled, arrived, completed, cancelled, no-show).
status_mapping = {
    "scheduled": "SCHEDULED",
    "arrived": "ARRIVED",
    "completed": "COMPLETED",
    "complete": "COMPLETED",
    "done": "COMPLETED",
    "cancelled": "CANCELLED",
    "canceled": "CANCELLED",
    "no-show": "NO-SHOW",
    "no show": "NO-SHOW"
}

for col in ["Status", "Appointment Status"]:
    if col in df.columns:
        df[col] = df[col].astype("string").str.strip().str.lower()
        df[col] = df[col].replace(status_mapping)

df[["Status", "Appointment Status"]].head(10)
```

```python
#6. Validate admission, transfer, procedure, and discharge event ordering; flag impossible sequences.
columns = ["Admission", "Transfer", "Procedure DateTime", "Discharge"]

for col in columns:
    df[col] = pd.to_datetime(df[col], errors="coerce", utc=True)

df["Event Order Issue"] = (
    (df["Admission"].notna() & df["Transfer"].notna() & (df["Transfer"] < df["Admission"])) |
    (df["Admission"].notna() & df["Procedure DateTime"].notna() & (df["Procedure DateTime"] < df["Admission"])) |
    (df["Admission"].notna() & df["Discharge"].notna() & (df["Discharge"] < df["Admission"])) |
    (df["Transfer"].notna() & df["Discharge"].notna() & (df["Discharge"] < df["Transfer"])) |
    (df["Procedure DateTime"].notna() & df["Discharge"].notna() & (df["Discharge"] < df["Procedure DateTime"]))
)

df[columns + ["Event Order Issue"]].head(10)
```

```python
#7. Normalize laboratory test names, specimen types, result units, and reference-range metadata using a catalogue.
lab_mapping = {
    "Lab Test": {"CBC": "COMPLETE BLOOD COUNT"},
    "Specimen Type": {"BLOOD": "Blood", "blood": "Blood"},
    "Lab Result Unit": {"MG/DL": "mg/dL", "MMOL/L": "mmol/L", "G/DL": "g/dL", "U/L": "U/L"},
    "Reference Range": {"NOT_APPLICABLE": None, "NOT APPLICABLE": None}
}

for col, mapping in lab_mapping.items():
    df[col] = df[col].astype("string").str.strip().replace(mapping)

df[["Lab Test", "Specimen Type", "Lab Result Unit", "Reference Range"]].head(10)
```

```python
#8. Convert compatible measurement units only when the source unit and conversion rule are known.
# Standardize the format of lab result units
df["Lab Result Unit"] = df["Lab Result Unit"].astype("string").str.strip()

df["Lab Result Unit"] = df["Lab Result Unit"].replace({
    "mmol/l": "mmol/L",
    "MMOL/L": "mmol/L",
    "MG/DL": "mg/dL",
    "G/DL": "g/dL"
})

# Display the standardized units
df["Lab Result Unit"].value_counts()
```

```python
#9. Standardize diagnosis and procedure codes to the specified coding-system version; flag invalid codes.
# Clean diagnosis and procedure codes
for col in ["Diagnosis Code", "Procedure Code"]:
    df[col] = df[col].astype("string").str.strip().str.upper()

    df[col + " Issue"] = (
        df[col].isna() |
        df[col].isin(["", "INVALID_CODE", "PROC-X", "NOT_APPLICABLE"])
    )

# Display the results
print(df[["Diagnosis Code", "Diagnosis Code Issue",
          "Procedure Code", "Procedure Code Issue"]].head(10))
```

```python
#10. Normalize medication administration status and dosage units; do not infer missing clinical values.
# Standardize medication administration status
df["Medication Administration Status"] = (
    df["Medication Administration Status"]
    .astype("string")
    .str.strip()
    .str.lower()
    .replace({
        "administered": "ADMINISTERED",
        "not administered": "NOT ADMINISTERED",
        "missed": "MISSED",
        "held": "HELD"
    })
)

# Standardize dosage unit format
df["Dosage Unit"] = (
    df["Dosage Unit"]
    .astype("string")
    .str.strip()
    .replace({
        "MG": "mg",
        "MCG": "mcg",
        "ML": "mL",
        "G": "g",
        "UNITS": "units"
    })
)

# Display the results
print(df[["Medication Administration Status", "Dosage Unit"]].head(10))
```

```python
#11. Deduplicate repeated appointment, lab, and bed-movement events using event keys and source priority.
# Remove duplicate appointment events
df = df.drop_duplicates(
    subset=["Appointment ID", "Patient ID", "Appointment DateTime"],
    keep="first"
)

# Remove duplicate lab events
df = df.drop_duplicates(
    subset=["Lab Order ID", "Encounter ID", "Lab Test", "Lab Result DateTime"],
    keep="first"
)

# Display the results
print("Records remaining after deduplication:", len(df))
print(df[["Appointment ID", "Lab Order ID", "Bed ID"]].head(10))
```

```python
#12. Validate bed, ward, and facility identifiers against the bed inventory.
# Check for missing bed, ward, and facility information
for col in ["Bed ID", "Ward", "Facility"]:
    df[col] = df[col].astype("string").str.strip()

df["Bed Information Issue"] = (
    df["Bed ID"].isna() | (df["Bed ID"] == "") |
    df["Ward"].isna() | (df["Ward"] == "") |
    df["Facility"].isna() | (df["Facility"] == "")
)

# Display the results
print(df[["Bed ID", "Ward", "Facility", "Bed Information Issue"]].head(10))
```

```python
#13. Standardize payer and billing categories; parse billed, paid, and outstanding amounts as decimals.
# Standardize payer names
df["Payer"] = df["Payer"].astype("string").str.strip().str.upper()

df["Payer"] = df["Payer"].replace({
    "SELF PAY": "SELF-PAY",
    "SELF-PAY": "SELF-PAY",
    "GOVT": "GOVERNMENT"
})

# Convert billing amounts to decimal numbers
amount_columns = ["Billed Amount", "Paid Amount", "Outstanding Amount"]

for col in amount_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce").astype("float64")

# Display the results
print(df[["Payer"] + amount_columns].head(10))
```

```python
#14. Normalize staff shift labels and date boundaries for overnight shifts.
# Standardize staff shift labels
df["Staff Shift"] = df["Staff Shift"].astype("string").str.strip().str.upper()

df["Staff Shift"] = df["Staff Shift"].replace({
    "NIGHT SHIFT": "NIGHT",
    "NIGHT-SHIFT": "NIGHT",
    "DAY SHIFT": "DAY",
    "DAY-SHIFT": "DAY"
})

# Convert shift start and end times
df["Shift Start"] = pd.to_datetime(df["Shift Start"], format="mixed", errors="coerce")
df["Shift End"] = pd.to_datetime(df["Shift End"], format="mixed", errors="coerce")

# Move the end time to the next day for overnight shifts
df.loc[df["Shift End"] < df["Shift Start"], "Shift End"] += pd.Timedelta(days=1)

# Display the results
df[["Staff Shift", "Shift Start", "Shift End"]].head(10)
```

```python
#15. Handle missing arrival, triage, or discharge timestamps using explicit null rules rather than fabricated values.
# Flag records with missing timestamps
df["Timestamp Issue"] = ""

df.loc[df["Arrival"].isna(), "Timestamp Issue"] = "Missing Arrival"
df.loc[df["Triage"].isna(), "Timestamp Issue"] += " Missing Triage"
df.loc[df["Discharge"].isna(), "Timestamp Issue"] += " Missing Discharge"

# Display the results
print(df[["Status", "Arrival", "Triage", "Discharge", "Timestamp Issue"]].head(10))

print("Records requiring timestamp review:", (df["Timestamp Issue"] != "").sum())
```

```python
#16. Flag negative wait durations, discharge before admission, and lab results before specimen collection.
#Flag negative wait durations (Triage before Arrival)
df["Negative Wait"] = (
    df["Arrival"].notna() &
    df["Triage"].notna() &
    (df["Triage"] < df["Arrival"])
)

#Flag discharge before admission
df["Discharge Before Admission"] = (
    df["Admission"].notna() &
    df["Discharge"].notna() &
    (df["Discharge"] < df["Admission"])
)

#Display flagged records
print("Records with Negative Wait:")
print(df[df["Negative Wait"]][["Encounter ID", "Arrival", "Triage"]].head(10))

print("\nRecords with Discharge Before Admission:")
print(df[df["Discharge Before Admission"]][["Encounter ID", "Admission", "Discharge"]].head(10))

# Display counts
print("\nTotal negative wait records:", df["Negative Wait"].sum())
print("Total discharge-before-admission records:", df["Discharge Before Admission"].sum())

print("\nNote: Lab result vs. specimen collection cannot be checked because a separate specimen collection timestamp is not available.")
```

```python
#17. Separate test records and cancelled encounters from operational denominators using documented criteria.
# Identify cancelled encounters
df["Cancelled Encounter"] = (
    df["Status"].astype("string").str.strip().str.upper() == "CANCELLED"
)

# Identify possible test records using Clinical Notes
df["Test Record"] = (
    df["Clinical Notes"].astype("string").str.strip().str.upper()
    .str.contains("TEST", na=False)
)

# Identify records included in operational calculations
df["Include in Operations"] = (
    ~df["Cancelled Encounter"] & ~df["Test Record"]
)

# Display the results
print(df[["Encounter ID", "Status", "Clinical Notes",
          "Cancelled Encounter", "Test Record",
          "Include in Operations"]].head(10))

print("\nCancelled encounters:", df["Cancelled Encounter"].sum())
print("Possible test records:", df["Test Record"].sum())
print("Records included in operations:", df["Include in Operations"].sum())
```

```python
#18. Normalize severity and priority labels only according to approved mapping tables.
# Clean Severity and Priority labels
for col in ["Severity", "Priority"]:
    df[col] = df[col].astype("string").str.strip().str.upper()

# Display the results
print(df[["Severity", "Priority"]].head(10))

# Show all unique labels
print("\nSeverity labels:", df["Severity"].unique())
print("Priority labels:", df["Priority"].unique())
```

```python
#19. Link lab orders and results using encounter/order keys and identify orphan results.
# Check for missing lab order or encounter IDs
df["Orphan Lab Result"] = (
    df["Lab Order ID"].isna() |
    df["Encounter ID"].isna() |
    df["Lab Order ID"].astype("string").str.strip().isin(["", "NOT_APPLICABLE"]) |
    df["Encounter ID"].astype("string").str.strip().isin(["", "NOT_APPLICABLE"])
)

# Check lab orders linked to multiple encounters
order_links = df.groupby("Lab Order ID")["Encounter ID"].nunique()
invalid_orders = order_links[order_links > 1]

# Display the results
print(df[["Lab Order ID", "Encounter ID", "Lab Test",
          "Lab Result", "Orphan Lab Result"]].head(10))

print("\nLab results missing a linking ID:", df["Orphan Lab Result"].sum())
print("Lab orders linked to multiple encounters:", len(invalid_orders))
print(invalid_orders)
```

```python
#20. Apply role-based masking to notes and remove names, phone numbers, and other direct identifiers.
# Mask clinical notes
df["Clinical Notes"] = "MASKED"

# Remove direct patient identifiers
df = df.drop(
    columns=["Patient ID", "Patient Name", "Phone"],
    errors="ignore"
)

# Display the privacy-safe dataset
print(df.head(10))

# Verify that identifiers have been removed
print("\nRemaining direct identifiers:",
      [col for col in ["Patient ID", "Patient Name", "Phone"]
       if col in df.columns])
```

```python
#Standardize text and missing-value placeholders
for col in df.columns:
    if df[col].dtype == "object" or isinstance(df[col].dtype, pd.StringDtype):
        df[col] = df[col].astype("string").str.strip()

        df[col] = df[col].replace({
            "": pd.NA,
            "NOT_APPLICABLE": pd.NA,
            "NOT APPLICABLE": pd.NA,
            "N/A": pd.NA,
            "NULL": pd.NA,
            "null": pd.NA
        })

#Standardize status columns
for col in ["Status", "Appointment Status", "Procedure Status",
            "Lab Result Status", "Medication Administration Status"]:
    if col in df.columns:
        df[col] = df[col].str.upper()

#Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

#Check duplicate encounter IDs
print("Duplicate Encounter IDs:",
      df["Encounter ID"].duplicated().sum())

#Show remaining missing values
print("\nMissing values by column:")
print(df.isnull().sum()[df.isnull().sum() > 0])

#Show final dataset size
print("\nFinal rows:", len(df))
print("Final columns:", len(df.columns))
print("\nCleaning checks completed.")
```

```python
#Standardize all column names
df.columns = (
    df.columns.str.strip()
    .str.replace(" ", "_")
    .str.replace(r"[^A-Za-z0-9_]", "", regex=True)
)

#Standardize text values
for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].astype("string").str.strip()

#Standardize status values
status_columns = [
    "Status",
    "Appointment_Status",
    "Procedure_Status",
    "Lab_Result_Status",
    "Medication_Administration_Status"
]

for col in status_columns:
    if col in df.columns:
        df[col] = df[col].str.upper()

# Standardize common status variations
if "Status" in df.columns:
    df["Status"] = df["Status"].replace({
        "COMPLETE": "COMPLETED",
        "CANCELED": "CANCELLED",
        "NO SHOW": "NO-SHOW"
    })

#Convert billing columns to numeric
amount_columns = [
    "Billed_Amount",
    "Paid_Amount",
    "Outstanding_Amount"
]

for col in amount_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

#Flag invalid event sequences
df["Negative_Wait"] = (
    df["Arrival"].notna() &
    df["Triage"].notna() &
    (df["Triage"] < df["Arrival"])
) if "Arrival" in df.columns and "Triage" in df.columns else False

df["Discharge_Before_Admission"] = (
    df["Admission"].notna() &
    df["Discharge"].notna() &
    (df["Discharge"] < df["Admission"])
) if "Admission" in df.columns and "Discharge" in df.columns else False

#Flag billing issues
if all(col in df.columns for col in amount_columns):
    df["Billing_Issue"] = (
        (df["Billed_Amount"] < 0) |
        (df["Paid_Amount"] < 0) |
        (df["Outstanding_Amount"] < 0) |
        (df["Paid_Amount"] > df["Billed_Amount"])
    )
else:
    df["Billing_Issue"] = False

#Create a privacy-safe dataset
df_safe = df.copy()

if "Clinical_Notes" in df_safe.columns:
    df_safe["Clinical_Notes"] = "MASKED"

df_safe = df_safe.drop(
    columns=["Patient_ID", "Patient_Name", "Phone"],
    errors="ignore"
)

#Create exception records
exception_mask = (
    df["Negative_Wait"] |
    df["Discharge_Before_Admission"] |
    df["Billing_Issue"]
)

exceptions = df.loc[exception_mask].copy()

#Save both CSV files
df_safe.to_csv("Hospital_Cleaned_Data.csv", index=False)
exceptions.to_csv("Hospital_Exception_Records.csv", index=False)

#Display final results
print("Cleaning and standardization completed!")
print("Total rows:", len(df_safe))
print("Total columns:", len(df_safe.columns))
print("Exception records:", len(exceptions))
print("Duplicate rows:", df_safe.duplicated().sum())
print("Column names:")
print(df_safe.columns.tolist())
```
